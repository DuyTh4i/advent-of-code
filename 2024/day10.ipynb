```python
dx = [1, 0, -1, 0]
dy = [0, 1, 0, -1]
ending = set()
map = []

def find_path(x, y, step):
    if step == 10:
        ending.add((x, y))
    for i in range(4):
        if 0 <= x + dx[i] < len(map) and 0 <= y + dy[i] < len(map) and map[x + dx[i]][y + dy[i]] == str(step):
            find_path(x + dx[i], y + dy[i], step + 1)

with open("day10.in", 'r') as file:
    map = file.read().splitlines()

    cnt = 0
    for x in range(len(map)):
        for y in range(len(map)):
            if map[x][y] == '0':
                ending.clear()
                find_path(x, y, 1)
                cnt += len(ending)
    print(cnt)
```

```python
dx = [1, 0, -1, 0]
dy = [0, 1, 0, -1]
path = set()
map = []

def find_path(x, y, step, history):
    if step == 10:
        path.add(history + str(x) + str(y))
    for i in range(4):
        if 0 <= x + dx[i] < len(map) and 0 <= y + dy[i] < len(map) and map[x + dx[i]][y + dy[i]] == str(step):
            find_path(x + dx[i], y + dy[i], step + 1, history + str(x) + str(y))

with open("day10.in", 'r') as file:
    map = file.read().splitlines()

    cnt = 0
    for x in range(len(map)):
        for y in range(len(map)):
            if map[x][y] == '0':
                find_path(x, y, 1, '')

    print(len(path))
```
