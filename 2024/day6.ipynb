```python
def check_valid_coor(x, y, edge_x, edge_y):
    return 0 <= x < edge_x and 0 <= y < edge_y

def solve(x, y, edge_x, edge_y, vertical, horizontal, matrix, visited):
    if not check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y):
        return len(visited)
    if matrix[x + vertical][y + horizontal] == "#":
        return solve(x, y, edge_x, edge_y, horizontal, vertical * -1 , matrix, visited)
    while check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y) and matrix[x + vertical][y + horizontal] != "#":
        x += vertical
        y += horizontal
        visited.add((x, y))
    return solve(x, y, edge_x, edge_y, vertical, horizontal, matrix, visited)

with open("day6.in","r") as file:
    matrix = list(map(list, file.read().split("\n")))
    visited = set()
    x,y = 0, 0
    for i in range(len(matrix)):
        if '^' in matrix[i]:
            x, y = i, matrix[i].index('^')
            visited.add((x, y))
            break
    print(solve(x, y, len(matrix), len(matrix[0]), -1, 0, matrix, visited))
```

```python
def check_valid_coor(x, y, edge_x, edge_y):
    return 0 <= x < edge_x and 0 <= y < edge_y

def check_loop(x, y, edge_x, edge_y, vertical, horizontal, matrix):
    check_loop_visited = set((x, y, vertical, horizontal))
    while check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y):
        if matrix[x + vertical][y + horizontal] == "#":
            vertical, horizontal = horizontal, vertical * -1
        else:
            x += vertical
            y += horizontal
            if (x, y, vertical, horizontal) in check_loop_visited:
                return True
            check_loop_visited.add((x, y, vertical, horizontal))
    return False

def get_visited(x, y, edge_x, edge_y, vertical, horizontal, matrix, visited):
    if not check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y):
        return visited
    if matrix[x + vertical][y + horizontal] == "#":
        return get_visited(x, y, edge_x, edge_y, horizontal, vertical * -1 , matrix, visited)
    while check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y) and matrix[x + vertical][y + horizontal] != "#":
        x += vertical
        y += horizontal
        visited.add((x, y))
    return get_visited(x, y, edge_x, edge_y, vertical, horizontal, matrix, visited)

with open("day6.in","r") as file:
    matrix = list(map(list, file.read().split("\n")))
    visited = set()
    x,y = 0, 0
    for i in range(len(matrix)):
        if '^' in matrix[i]:
            x, y = i, matrix[i].index('^')
            visited.add((x, y))
            break
    visited = get_visited(x, y, len(matrix), len(matrix[0]), -1, 0, matrix, visited)
    visited.remove((x, y))
    result = 0
    for position in visited:
        matrix[position[0]][position[1]] = '#'
        if check_loop(x, y, len(matrix), len(matrix[0]), -1, 0, matrix):
            result += 1
        matrix[position[0]][position[1]] = '.'
    print(result)
```
