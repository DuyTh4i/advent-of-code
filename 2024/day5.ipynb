```python
with open("fifth.in", "r") as file:
    rules = {}
    result = 0
    for line in file.readlines():
        if "|" in line:
            x, y = line.strip().split("|")
            if x not in rules:
                rules[x] = [y]
            else: rules[x].append(y)
        elif "," in line:
            nums =  line.strip().split(",")
            ok = True
            for i in range(len(nums)):
                for j in range(i+1, len(nums)):
                    if nums[j] in rules and nums[i] in rules[nums[j]]:
                        ok = False
                        break
                if not ok:
                    break
            if ok:
                result += int(nums[int((len(nums))/2)])
    print (result)
```
