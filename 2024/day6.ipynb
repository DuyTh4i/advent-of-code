```python
def check_valid_coor(x, y, edge_x, edge_y):
    return x >= 0 and y >= 0 and y < edge_y and x < edge_x

def solve(x, y, edge_x, edge_y, vertical, horizontal, cnt, matrix, matrix_visit):
    if x + vertical < 0 or y + horizontal < 0 or y + horizontal >= edge_y or x + vertical >= edge_x:
        return cnt
    if matrix[x + vertical][y + horizontal] == "#":
        return solve(x, y, edge_x, edge_y, horizontal, vertical * -1 , cnt, matrix, matrix_visit)
    while check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y) and matrix[x + vertical][y + horizontal] != "#":
        x += vertical
        y += horizontal
        if matrix_visit[x][y] == 0:
            matrix_visit[x][y] = 1
            cnt += 1
    return solve(x, y, edge_x, edge_y, vertical, horizontal, cnt, matrix, matrix_visit)

with open("day6.in","r") as file:
    matrix = file.read().split("\n")
    matrix_visit = [[0 for _ in range(len(matrix[0]))] for _ in range(len(matrix))]
    x,y = 0, 0
    for i in range(len(matrix)):
        if "^" in matrix[i]:
            x, y = i, matrix[i].find("^")
            matrix_visit[x][y] = 1
            break
    print(solve(x, y, len(matrix), len(matrix[0]), -1, 0, 1, matrix, matrix_visit))
```

```python
def check_valid_coor(x, y, edge_x, edge_y):
    return x >= 0 and y >= 0 and y < edge_y and x < edge_x

def check_loop(x, y, edge_x, edge_y, vertical, horizontal, matrix, turn_points):
    while check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y):
        if turn_points[x + vertical][y + horizontal] == 1 and matrix[x + vertical*2][y + horizontal*2] == "#":
            return True
        x += vertical
        y += horizontal
    return False

def solve(x, y, edge_x, edge_y, vertical, horizontal, cnt, matrix, turn_points):
    if x + vertical < 0 or y + horizontal < 0 or y + horizontal >= edge_y or x + vertical >= edge_x:
        return cnt
    if matrix[x + vertical][y + horizontal] == "#":
        turn_points[x][y] = 1
        return solve(x, y, edge_x, edge_y, horizontal, vertical * -1 , cnt, matrix, turn_points)
    while check_valid_coor(x + vertical, y + horizontal, edge_x, edge_y) and matrix[x + vertical][y + horizontal] != "#":
        x += vertical
        y += horizontal
        if check_loop(x, y, edge_x, edge_y, horizontal, vertical * -1 , matrix, turn_points):
            cnt += 1
    return solve(x, y, edge_x, edge_y, vertical, horizontal, cnt, matrix, turn_points)

with open("day6.in","r") as file:
    matrix = file.read().split("\n")
    turn_points = [[0 for _ in range(len(matrix[0]))] for _ in range(len(matrix))]
    x,y = 0, 0
    for i in range(len(matrix)):
        if "^" in matrix[i]:
            x, y = i, matrix[i].find("^")
            break
    print(solve(x, y, len(matrix), len(matrix[0]), -1, 0, 0, matrix, turn_points))
```
