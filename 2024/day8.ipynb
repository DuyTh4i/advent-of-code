```python
def is_valid_coor(x, y, edge_x, edge_y):
    return 0 <= x < edge_x and 0 <= y < edge_y

with open("day8.in", "r") as data:
    result = set()
    antennas_list = {}
    matrix = data.read().splitlines()
    ans = 0
    for x, line in enumerate(matrix):
        for y in range(len(line)):
            if line[y] != '.':
                result.add((x, y))
                if line[y] not in antennas_list:
                    antennas_list[line[y]] = [(x, y)]
                else: antennas_list[line[y]].append((x, y))

    for pos_list in antennas_list.values():
        for pivot in pos_list:
            for pos in pos_list:
                if pivot == pos: continue
                x, y = pivot[0], pivot[1]
                diff_x = pivot[0] - pos[0]
                diff_y = pivot[1] - pos[1]
                while is_valid_coor(x + diff_x, y + diff_y, len(matrix), len(matrix[0])):
                    x += diff_x
                    y += diff_y
                    result.add((x, y))
    print(len(result))
```
