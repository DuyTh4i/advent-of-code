https://adventofcode.com/2024/day/6

```python
def solve(x, y, edge_x, edge_y, vertical, horizontal, cnt, matrix, matrix_visit):
    if x + vertical < 0 or y + horizontal < 0 or y + horizontal >= edge_y or x + vertical >= edge_x:
        return cnt
    if matrix[x + vertical][y + horizontal] == "#":
        return solve(x, y, edge_x, edge_y, horizontal, vertical * -1 , cnt, matrix, matrix_visit)
    while x + vertical >= 0 and y + horizontal >= 0 and y + horizontal < edge_y and x + vertical < edge_x and matrix[x + vertical][y + horizontal] != "#":
        x += vertical
        y += horizontal
        if matrix_visit[x][y] == 0:
            matrix_visit[x][y] = 1
            cnt += 1
    return solve(x, y, edge_x, edge_y, vertical, horizontal, cnt, matrix, matrix_visit)

with open("day6.in","r") as file:
    matrix = file.read().split("\n")
    matrix_visit = [[0 for _ in range(len(matrix[0]))] for _ in range(len(matrix))]
    x,y = 0, 0
    for i in range(len(matrix)):
        if "^" in matrix[i]:
            x, y = i, matrix[i].find("^")
            matrix_visit[x][y] = 1
            break
    print(solve(x, y, len(matrix), len(matrix[0]), -1, 0, 1, matrix, matrix_visit))
```
