```python
def decode_map(disk_map):
    decoded_map = []
    id = 0
    for index, element in enumerate(disk_map):
        data = '.'
        if not index % 2:
            data = str(id)
            id += 1
        for i in range(int(element)):
            decoded_map.append(data)
    return decoded_map

def solve(decoded_map):
    start, end = 0, (len(decoded_map) - 1)
    ans = 0
    while start <= end:
        if decoded_map[start] == '.':
            while decoded_map[end] == '.':
                end -= 1
            ans += start * int(decoded_map[end])
            end -= 1
            start += 1
        else: 
            ans += start * int(decoded_map[start])
            start += 1
    return ans

with open("day9.in", "r") as input:
    disk_map = input.readline()
    print(solve(decode_map(disk_map)))
```

```python
def decode_map(disk_map):
    folders_map = []
    dots_map = []
    id = 0
    start = 0
    for index, element in enumerate(disk_map):
        if not index % 2:
            folders_map.append([start, int(element), id])
            id += 1
        else: dots_map.append([start, int(element)])
        start += int(element)
    return folders_map, dots_map

def solve(folders_map, dots_map):
    for folder in folders_map[::-1]:
        for dot in dots_map:
            if dot[0] < folder[0] and dot[1] >= folder[1]:
                folder[0] = dot[0]
                dot[0] += folder[1]
                dot[1] = dot[1] - folder[1]
    sorted_folders_map = sorted(folders_map, key=lambda x: x[0])
    
    result = 0
    for folder in sorted_folders_map:
        for i in range(folder[0], folder[0] + folder[1]):
            result += i * folder[2]
    return result

with open("day9.in", "r") as input:
    disk_map = input.readline()
    folders_map, dots_map = decode_map(disk_map)
    print(solve(folders_map, dots_map))
```
