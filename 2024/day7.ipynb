```python
def solve(total, array, index, temp, answer):
    if temp > total:
        return 0

    if index == len(array):
        return total == temp
    
    answer += solve(total, array, index + 1, temp+array[index], answer)
    answer += solve(total, array, index + 1, temp*array[index], answer)

    return answer

with open("day7.in", 'r') as file:
    for line in file.read().split('\n'):
        input = line.split(": ")
        total = int(input[0])
        array = list(map(int, input[1].strip().split(' ')))
    print(solve(total, array, 1, array[0], 0))
        
            
            
        
```
