```python
dp = {}

def blink(rock, times):
    if (rock, times) in dp: return dp[(rock, times)]
    if not rock: return 0
    if not times: return 1

    left, right = '', ''

    if rock == '0':
        left = '1'
    elif len(rock) % 2 == 0:
        right = str(int(rock[int(len(rock)/2) : ]))
        left = str(int(rock[ : int(len(rock)/2)]))
    else: left= str(int(rock) * 2024)

    dp[(rock, times)] = blink(left, times - 1) + blink(right, times - 1)
    return dp[(rock, times)]

with open('day11.in', 'r') as file:
    rocks = file.read().split(' ')
    cnt = 0

    for rock in rocks:
        cnt += blink(rock, 25)

    print(cnt)
```
