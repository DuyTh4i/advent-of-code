```python
def solve(total, array, index, temp):
    if temp > total: return False
    if index == len(array): return total == temp
    if solve(total, array, index + 1, temp+array[index]): return True
    return solve(total, array, index + 1, temp*array[index])

with open("day7.in", 'r') as file:
    result = 0
    for line in file.read().split('\n'):
        input = line.split(": ")
        total = int(input[0])
        array = list(map(int, input[1].strip().split(' ')))
        if solve(total, array, 1, array[0]):
            result += total
    print(result)
```

```python
def solve(total, array, index, temp):
    if temp > total: return False
    if index == len(array): return total == temp
    if solve(total, array, index + 1, temp+array[index]):
        return True
    if solve(total, array, index + 1, temp*array[index]):
        return True
    if solve(total, array, index + 1, int(str(temp) + str(array[index]))):
        return True
    return False

with open("day7.in", 'r') as file:
    result = 0
    for line in file.read().split('\n'):
        input = line.split(": ")
        total = int(input[0])
        array = list(map(int, input[1].strip().split(' ')))
        if solve(total, array, 1, array[0]):
            result += total
    print(result)
```
